### Additional Visualizations

```python
top_countries = df['Country'].value_counts().nlargest(10).index
df_top_countries = df[df['Country'].isin(top_countries)]

plt.figure(figsize=(12, 7))
sns.boxplot(x='Overall_ESG_Score', y='Country', data=df_top_countries, hue='Country', palette='viridis', legend=False)
plt.title('Overall ESG Score Distribution by Top 10 Countries')
plt.xlabel('Overall ESG Score')
plt.ylabel('Country')
plt.grid(axis='x', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
avg_emp_satisfaction_by_industry = df.groupby('Industry')['Employee_Satisfaction_Score'].mean().sort_values(ascending=False).reset_index()

plt.figure(figsize=(12, 7))
sns.barplot(x='Employee_Satisfaction_Score', y='Industry', data=avg_emp_satisfaction_by_industry, hue='Industry', palette='magma', legend=False)
plt.title('Average Employee Satisfaction Score by Industry')
plt.xlabel('Average Employee Satisfaction Score')
plt.ylabel('Industry')
plt.grid(axis='x', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
# Convert 'Reporting_Year' to numeric, coercing errors
df['Reporting_Year_Numeric'] = pd.to_numeric(df['Reporting_Year'], errors='coerce')

# Fill NaN values in 'Reporting_Year_Numeric' with the median year for plotting trends
median_year = df['Reporting_Year_Numeric'].median()
df['Reporting_Year_Numeric'] = df['Reporting_Year_Numeric'].fillna(median_year).astype(int)

# Group by the numeric reporting year and calculate the mean of carbon emissions
avg_carbon_emissions_by_year = df.groupby('Reporting_Year_Numeric')['Carbon_Emissions_tCO2'].mean().reset_index()

plt.figure(figsize=(12, 6))
sns.lineplot(x='Reporting_Year_Numeric', y='Carbon_Emissions_tCO2', data=avg_carbon_emissions_by_year, marker='o') # Removed palette since hue is not assigned for a single line
plt.title('Average Carbon Emissions Over Reporting Years')
plt.xlabel('Reporting Year')
plt.ylabel('Average Carbon Emissions (tCO2)')
plt.grid(True, linestyle='--', alpha=0.6)
plt.xticks(rotation=45) # Rotate x-axis labels if needed for readability
plt.tight_layout()
plt.show()
```
